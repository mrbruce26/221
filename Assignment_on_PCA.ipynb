```python
import pandas as pd
import urllib.request
import gzip

url = "https://ftp.ncbi.nlm.nih.gov/geo/series/GSE309nnn/GSE309664/suppl/GSE309664_counts.txt.gz"
filename = "GSE309664_counts.txt.gz"


print(f"Downloading {filename}...")
urllib.request.urlretrieve(url, filename)
print("Download complete.")

expression_df = pd.read_csv(filename, sep='\t', index_col=0)

print(f"\nSUCCESS! Expression matrix shape: {expression_df.shape}")
print("\nFirst few rows:")
print(expression_df.head())
```

```python

expression_df.to_csv('GSE309664_Expression_Matrix.tsv', sep='\t', index=True)
print("\n Expression data saved as: GSE309664_Expression_Matrix.tsv")
```

```python
filename = "GSE309664_Expression_Matrix.tsv"
loaded_expression_df  = pd.read_csv(filename,sep = '\t',index_col =0)

loaded_expression_df.head()
```

```python
X_data = expression_df.T
```

```python
X_data.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_data)
```

```python

from sklearn.decomposition import PCA
pca = PCA(n_components=0.95)
principal_components = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(data = principal_components)


print(f"Original dimension (Genes): {X_scaled.shape[1]}")
print(f"Reduced dimension (PCs needed for 95% variance): {pca.n_components_}")
print(f"Total variance explained: {pca.explained_variance_ratio_.sum():.4f}")
```

```python
loadings_matrix = pca.components_

loadings_df = pd.DataFrame(
    data=loadings_matrix.T,
    index=X_data.columns,
    columns=[f'PC{i+1}' for i in range(pca.n_components_)]
)

print(f" Loadings DataFrame created with shape: {loadings_df.shape}")
print("\nLoadings Head (First 5 Genes across PC1-PC5):")
print(loadings_df.iloc[:, :5].head())
```

```python
print(principal_components[0])
```

```python

top_5_loadings_df = loadings_df.iloc[:, :5]

variance_ratios = pca.explained_variance_ratio_
variance_series = pd.Series(
    variance_ratios[:5],
    index=[f'PC{i+1}' for i in range(5)]
)

abs_loadings = top_5_loadings_df.abs()

weighted_abs_loadings = abs_loadings.mul(variance_series, axis='columns')

weighted_score = weighted_abs_loadings.sum(axis=1)


top_20_weighted_genes = weighted_score.sort_values(ascending=False).head(20)

print("--- Top 20 Genes Based on Overall Weighted Influence (PC1-PC5) ---")
print(top_20_weighted_genes)
```

```python
loadings_matrix = pca.components_

loadings_df = pd.DataFrame(
    data=loadings_matrix.T,
    index=X_data.columns,
    columns=[f'PC{i+1}' for i in range(pca.n_components_)]
)

print(f" Loadings DataFrame created with shape: {loadings_df.shape}")
print("\nLoadings Head (First 5 Genes across PC1-PC10):")
print(loadings_df.iloc[:, :10].head())
```

```python

top_5_loadings_df = loadings_df.iloc[:, :10]

variance_ratios = pca.explained_variance_ratio_
variance_series = pd.Series(
    variance_ratios[:10],
    index=[f'PC{i+1}' for i in range(10)]
)

abs_loadings = top_5_loadings_df.abs()

weighted_abs_loadings = abs_loadings.mul(variance_series, axis='columns')

weighted_score = weighted_abs_loadings.sum(axis=1)

top_20_weighted_genes = weighted_score.sort_values(ascending=False).head(20)

print("--- Top 20 Genes Based on Overall Weighted Influence (PC1-PC5) ---")
print(top_20_weighted_genes)
```

We can see that the plot with 10 principal components shows higher gene loading values compared to the one with 5 principal components. This happens because the original gene expression data (containing ~95% variance) was reduced into 33 principal components, meaning these components retain most of the informational structure of the dataset.

When we select only the first 5 PCs, we capture a smaller portion of this variance. Since the earliest PCs explain the most variation, choosing just 5 means we are working with a more compressed representation of the data. However, when we increase the selection to 10 PCs, we include more information from the dataset — so the overall contribution of genes becomes larger.

In PCA, loadings represent how much each gene contributes to a given principal component. Therefore, when we include more PCs, the total contribution values of genes increase, which is why the second list shows larger numbers than the first.

Using PCA for gene selection can be useful, but it has both strengths and limitations. PCA helps reduce dimensionality by transforming thousands of genes into a smaller number of principal components that capture the largest proportion of variance in the data. This is effective for identifying major expression patterns and removing noise, making downstream analysis faster and more manageable. Selecting genes based on high loadings can therefore highlight those that contribute strongly to the major sources of variance in the dataset.
