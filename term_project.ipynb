```python
import requests
import gzip
import shutil

# 1. The URL for one specific chicken sample (BA-C1)
url = "ftp://ftp.sra.ebi.ac.uk/vol1/fastq/ERR118/009/ERR11836729/ERR11836729_1.fastq.gz"
output_filename = "chicken_subset.fastq"
line_limit = 100000  # 100,000 lines = 25,000 bacteria sequences

print(f"Starting stream from: {url}")
print(f"Saving first {line_limit} lines to {output_filename}...")

# 2. Open the connection with stream=True (Don't download all at once)
try:
    with requests.get(url, stream=True) as r:
        r.raise_for_status()

        # 3. Handle the GZIP compression on the fly
        # We wrap the raw stream in a GzipFile reader
        with gzip.open(r.raw, 'rt') as f_in, open(output_filename, 'w') as f_out:
            count = 0
            for line in f_in:
                f_out.write(line)
                count += 1
                if count >= line_limit:
                    break  # STOP downloading here!

    print(f"✅ Success! Saved {count} lines. You can now analyze '{output_filename}'.")

except Exception as e:
    print(f"❌ Error: {e}")
    print("Note: If the FTP link fails, the server might be busy. Try again in 5 mins.")
```
